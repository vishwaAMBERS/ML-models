```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

# Dataset with 20 data points forming 3 clusters
X = np.array([
[1,2],[1.5,1.8],[2,2],[2.2,2.5],[1.8,1.5],[2.5,2.2],[1.2,1.6],
[6,7],[6.5,7.5],[7,8],[7.2,7.8],[6.8,7.2],[7.5,8.2],[6.2,7.1],
[10,2],[10.5,2.5],[11,3],[10.8,2.8],[11.2,3.2],[10.3,2.2]
])

# Apply Agglomerative Hierarchical Clustering
model = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = model.fit_predict(X)

# Scatter plot of clusters
plt.scatter(X[:,0], X[:,1], c=labels)
plt.title("Hierarchical Clustering (20 Points, 3 Clusters)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

# Create dendrogram
linked = linkage(X, 'ward')

plt.figure(figsize=(8,5))
dendrogram(linked)
plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()

# -------------------------------
# Evaluation Metrics
# -------------------------------

sil_score = silhouette_score(X, labels)
db_score = davies_bouldin_score(X, labels)
ch_score = calinski_harabasz_score(X, labels)

print("Silhouette Score:", sil_score)
print("Davies-Bouldin Index:", db_score)
print("Calinski-Harabasz Index:", ch_score)
```
