```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

dataset = pd.read_csv("/content/drive/MyDrive/FeatureEnggData.csv")
print(dataset)
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

dataset_for_imputation = dataset.copy(deep=True)
x_impute = dataset_for_imputation.iloc[:, :-1].values # Exclude the target column 'Purchased'

# Identify numerical columns for median imputation ('Age' and 'Salary')
# Based on the dataset structure, 'Age' is at index 3 and 'Salary' at index 4 in the x_impute array
numerical_cols_indices = [3, 4]

# Create and apply the SimpleImputer with median strategy to numerical columns
imputer = SimpleImputer(missing_values=np.nan, strategy="median")
x_impute[:, numerical_cols_indices] = imputer.fit_transform(x_impute[:, numerical_cols_indices])

# Display the results of the imputation (first 5 rows for Age and Salary)
print("Imputed Age and Salary columns (first 5 rows):\n", x_impute[:5, numerical_cols_indices])

# To visualize the full dataset after imputation, create a DataFrame
columns_x = ['Name', 'Country', 'Gender', 'Age', 'Salary']
imputed_df = pd.DataFrame(x_impute, columns=columns_x)
print("\nDataFrame after median imputation for Age and Salary (first 5 rows):\n", imputed_df.head())
```

```python
threshold = 0.6
dataset = dataset[dataset.columns[dataset.isnull().mean() < threshold]]
print(dataset)
```

```python
from sklearn.impute import SimpleImputer
dataset2 = dataset.copy(deep=True)
x = dataset2.iloc[:, :-1].values

print("Median Age:", dataset2['Age'].median())
print("Median Salary:", dataset2['Salary'].median())

imputer = SimpleImputer(missing_values=np.nan, strategy="median")
# Apply imputer to 'Age' (index 3) and 'Salary' (index 4) columns
imputer.fit(x[:, 3:5])
x[:, 3:5] = imputer.transform(x[:, 3:5])

# Display the relevant columns after imputation
print(pd.DataFrame(x[:, [1, 2, 3, 4]], columns=['Country', 'Gender', 'Age', 'Salary']))
```

```python
dataset2 = dataset.copy(deep=True)
x = dataset2.iloc[:, :-1].values

print("Median Age:", dataset2['Age'].median())
print("Median Salary:", dataset2['Salary'].median())

imputer = SimpleImputer(missing_values=np.nan, strategy="median")
# Apply imputer only to 'Age' (index 3) and 'Salary' (index 4) columns
imputer.fit(x[:, 3:5])
x[:, 3:5] = imputer.transform(x[:, 3:5])

# Display the relevant columns after imputation: Country, Gender, Age, Salary
print(pd.DataFrame(x[:, 1:5], columns=['Country', 'Gender', 'Age', 'Salary']))
```
